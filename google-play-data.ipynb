# Cleaning the data

```python
#viewing the corrupted row
df[df['Category'] == '1.9']
```

#### Dropping the corrupted row (turns out all corrupted data are in one row)

```python
# removing corrupted Category value
df = df[df['Category'] != '1.9'].copy()
```

#### Handling duplicates

##### 1. Dropping identical duplicates

```python
df = df.drop_duplicates()
```

##### 2. Handling duplicated apps

```python
# we keep the row with the highest reviews count
df = df.sort_values('Reviews', ascending= False)

df = df.drop_duplicates(subset='App', keep='first')
```

```python
df = df.reset_index(drop=True)
```

```python
df
```

#### Converting installs to numeric

```python
df['Installs'] = df['Installs'].str.replace('+','').str.replace(',','').astype(int)
```

#### Coverting reviews to numeric

```python
df['Reviews'] = df['Reviews'].astype(int)
```

#### Converting price to numeric

```python
df['Price'] = df['Price'].str.replace('$','').astype(float)
```

#### Converting size to numeric (in MB)

```python
def convert_size_to_mb(value):
    """Convert a single Size value to a float in MB."""
    if value == 'Varies with device':
        return np.nan
    if value.endswith('M'):
        return float(value[:-1])
    if value.endswith('k'):
        return float(value[:-1]) / 1024
    return np.nan
```

```python
df['Size'] = df['Size'].apply(convert_size_to_mb)
```

```python
df['Size'] = df['Size'].round(3)
df['Size']
```

```python
df.info()
```

```python
df
```

# Handling missing values

```python
df.isna().mean()*100
```

#### dropping missing values rows for 'Current Ver' and 'Android Ver' and 'Type' columns (missing values percentages too low)

```python
df = df.dropna(subset=['Current Ver', 'Android Ver', 'Type'], ignore_index=True) 
df.isna().mean()*100
```

#### to fix pandas SettingWithCopyWarning, we create a copy of the DataFrame before filling missing values

```python
df = df.copy()
```

#### filling missing values for 'Rating' column with the median value of ratings for each category

```python
df['Rating'] = df.groupby('Category')['Rating'].transform(lambda r: r.fillna(r.median()))
```

```python
df.isna().mean()*100
```

#### missing values for 'Size' column won't be filled as it was 'Varies with device' so we add a new column 'size_varies' to indicate if the size varies with device or not

```python
df['size_varies'] = df['Size'].isna()
df
```

```python
df['size_varies'].value_counts()
```

#### Converting 'Last Updated' to date time

```python
df['Last Updated'] = pd.to_datetime(df['Last Updated'])
```
