# Cleaning the data

```python
#viewing the corrupted row
df[df['Category'] == '1.9']
```

#### Dropping the corrupted row (turns out all corrupted data are in one row)

```python
# removing corrupted Category value
df = df[df['Category'] != '1.9'].copy()
```

#### Handling duplicates

##### 1. Dropping identical duplicates

```python
df = df.drop_duplicates()
```

##### 2. Handling duplicated apps

```python
# we keep the row with the highest reviews count
df = df.sort_values('Reviews', ascending= False)

df = df.drop_duplicates(subset='App', keep='first')
```

```python
df = df.reset_index(drop=True)
```

```python
df
```

#### Converting installs to numeric

```python
df['Installs'] = df['Installs'].str.replace('+','').str.replace(',','').astype(int)
```

#### Coverting reviews to numeric

```python
df['Reviews'] = df['Reviews'].astype(int)
```

#### Converting price to numeric

```python
df['Price'] = df['Price'].str.replace('$','').astype(float)
```

#### Converting size to numeric (in MB)

```python
def convert_size_to_mb(value):
    """Convert a single Size value to a float in MB."""
    if value == 'Varies with device':
        return np.nan
    if value.endswith('M'):
        return float(value[:-1])
    if value.endswith('k'):
        return float(value[:-1]) / 1024
    return np.nan
```

```python
df['Size'] = df['Size'].apply(convert_size_to_mb)
```

```python
df['Size'] = df['Size'].round(3)
df['Size']
```

```python
df.info()
```

```python
df
```

# Handling missing values

```python
df.isna().mean()*100
```

#### dropping missing values rows for 'Current Ver' and 'Android Ver' and 'Type' columns (missing values percentages too low)

```python
df = df.dropna(subset=['Current Ver', 'Android Ver', 'Type'], ignore_index=True) 
df.isna().mean()*100
```

#### to fix pandas SettingWithCopyWarning, we create a copy of the DataFrame before filling missing values

```python
df = df.copy()
```

#### filling missing values for 'Rating' column with the median value of ratings for each category

```python
df['Rating'] = df.groupby('Category')['Rating'].transform(lambda r: r.fillna(r.median()))
```

```python
df.isna().mean()*100
```

#### missing values for 'Size' column won't be filled as it was 'Varies with device' so we add a new column 'size_varies' to indicate if the size varies with device or not

```python
df['size_varies'] = df['Size'].isna()
df
```

```python
df['size_varies'].value_counts()
```

#### Converting 'Last Updated' to date time

```python
df['Last Updated'] = pd.to_datetime(df['Last Updated'])
```

```python
df
```

# Exploration phase

#### What does the overall distribution of app ratings look like?

```python
fig = px.histogram(df, x='Rating', nbins=20, title='Distribution of App Ratings', labels={'Rating': 'Rating (1-5)'})

fig.show()
```

#### How are app sizes distributed, and does that differ between Free and Paid apps?

```python
fig = px.histogram(
     df, x='Size', color='Type', barmode='overlay', 
        nbins=40, title='App Size Distribution: Free vs Paid',
       labels={'Size': 'Size (MB)'}
       )

fig.show()
```

#### Which categories have the most apps available?

```python
category_counts = df['Category'].value_counts().head(15).reset_index()
fig = px.bar(category_counts, x = 'Category', y = 'count', title='Top 15 Categories by Number of Apps')
fig
```

#### Do certain categories tend to be rated higher than others?

```python
avg_rating_by_category = df.groupby('Category')['Rating'].mean().sort_values(ascending=False).reset_index()

fig = px.bar(avg_rating_by_category, x='Category', y='Rating', title='Average Rating by Category')

fig.show()
```

#### Which categories have the most installs overall?

```python
installs_by_category = df.groupby('Category')['Installs'].sum().sort_values(ascending=False).head(10).reset_index()

fig = px.bar(installs_by_category, x='Category', y='Installs', title='Top 10 Categories by Total Installs')

fig.show()
```

#### What share of apps are Free vs Paid?

```python
type_counts = df['Type'].value_counts().reset_index()
fig = px.pie(type_counts, names='Type', values='count', title='Free vs Paid Apps')

fig.show()
```

#### How is the app audience (Content Rating) split across the store?

```python
content_rating_counts = df['Content Rating'].value_counts().reset_index()

fig = px.pie(content_rating_counts, names='Content Rating', values='count', title='Apps by Content Rating Audience')

fig.show()
```

#### Do paid apps tend to be rated higher than free apps?

```python
fig = px.box(df, x='Type', y='Rating',title='Rating Distribution: Free vs Paid Apps',labels={'Rating': 'Rating (1-5)'})

fig.show()
```

#### How do the numeric variables in this dataset relate to each other overall?

```python
numeric_cols = ['Rating', 'Reviews', 'Size', 'Installs', 'Price']
corr_matrix = df[numeric_cols].corr()
fig = px.imshow(corr_matrix,text_auto= '.2f' ,title='Correlation Between Numeric Variables')

fig.show()
```
